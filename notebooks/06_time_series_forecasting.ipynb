# Phase 6: Time-Series Forecasting

In this phase, we explore how to predict future sensor behaviors using the ARIMA model, which is a fundamental technique for monitoring industrial systems over time.

```python
import pandas as pd
import matplotlib.pyplot as plt
import joblib
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error

# Load dataset
df = pd.read_csv("../datasets/predictive_maintenance.csv")
print("Dataset loaded.")
```

### Step 69 & 70: Select and Plot Sensor Signal

```python
temperature_series = df["Air temperature"]

plt.figure(figsize=(12,5))
plt.plot(temperature_series.iloc[:1000]) # Plotting first 1000 for clarity
plt.title("Air Temperature Over Time (First 1000 records)")
plt.xlabel("Time")
plt.ylabel("Temperature")
plt.show()
```

### Step 71: Train/Test Split for Time-Series

```python
# Time-series split (no shuffling)
train_size = int(len(temperature_series) * 0.8)
train = temperature_series[:train_size]
test = temperature_series[train_size:]

print(f"Training set size: {len(train)}")
print(f"Testing set size: {len(test)}")
```

### Step 72 & 73: ARIMA Model Training

```python
# Using a subset for faster training demonstration if needed,
# but here we follow the micro-steps.
model = ARIMA(train, order=(5,1,0))
model_fit = model.fit()
print("ARIMA model trained successfully.")
```

### Step 74 & 75: Forecast & Visualization

```python
forecast = model_fit.forecast(steps=len(test))

plt.figure(figsize=(12,5))
plt.plot(test.values[:100], label="Actual") # Plotting first 100 for clarity
plt.plot(forecast.values[:100], label="Forecast")
plt.legend()
plt.title("Temperature Forecast (First 100 test records)")
plt.show()
```

### Step 76: Forecast Error

```python
mae = mean_absolute_error(test, forecast)
print(f"Mean Absolute Error (MAE): {mae:.4f}")
```

### Step 77 & 78: Failure Trend Analysis

```python
failure_signal = df["Machine failure"].rolling(window=20).mean()

plt.figure(figsize=(12,5))
plt.plot(failure_signal)
plt.title("Failure Trend Signal (Rolling Mean of Failures)")
plt.show()
```

### Step 79: Save Forecast Model

```python
joblib.dump(model_fit, "../models/arima_forecast.pkl")
print("Forecasting model saved to ../models/arima_forecast.pkl")
```
